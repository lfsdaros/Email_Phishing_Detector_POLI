## 5. Análise de valores ausentes

Aqui vamos verificar se existem valores faltantes na base.

```python
# Quantidade de valores ausentes por coluna
missing = df.isnull().sum().sort_values(ascending=False)
missing_percent = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_table = pd.DataFrame({
    'valores_ausentes': missing,
    'percentual_%': missing_percent.round(2)
})

missing_table[missing_table['valores_ausentes'] > 0]
# Printando a tabela para ficar mais visível o fato de que não há valores nulos nas colunas
missing_table
```

```python
# Gráfico de valores ausentes, se existirem
missing_nonzero = missing_table[missing_table['valores_ausentes'] > 0]

if len(missing_nonzero) > 0:
    missing_nonzero['percentual_%'].plot(kind='bar', figsize=(12, 5))
    plt.title('Percentual de valores ausentes por coluna')
    plt.ylabel('Percentual (%)')
    plt.xlabel('Colunas')
    plt.xticks(rotation=45, ha='right')
    plt.show()
else:
    print('Não foram encontrados valores ausentes na base.')
```
